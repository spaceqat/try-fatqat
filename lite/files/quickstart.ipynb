# Your first quantum circuit

Create two entangled qubits, run the circuit, and compare their measurements.

> Select each code cell and press **Shift + Enter**. The first run may take a minute while Python and FatQat load.

## 1. Build a Bell pair

Two gates prepare the pair; measurements read out the two bits. Run this cell to see the circuit.

```python
import fatqat as fq
import fatqat.operations as ops

program = fq.Program(2, 2)
program.add(ops.H, 0)
program.add(ops.CX, (0, 1))
program.measure_all()

figure = program.draw()
```

## 2. Run the circuit

Repeat the experiment 1,000 times. Each run gives a pair of bits.

```python
backend = fq.simulator.Simulator(runtime="numpy")
result = backend.run(program, shots=1000).result()
```

## 3. Compare the outcomes

Expect roughly half `00` and half `11`. The two measured bits always agree for this ideal Bell pair.

```python
result.draw(title="One pair. Matching outcomes.")
```
